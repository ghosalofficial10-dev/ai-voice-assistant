# 🐱 Kitty — AI Assistant

A beginner-friendly, keyword-based chatbot built with Python. Run the cells from top to bottom, then try the sample messages.

```python
import re

def kitty_chat(user_input):
    user_input = user_input.lower().strip()

    if re.search(r"\b(hello|hi)\b", user_input):
        return "Hello! I'm Kitty 🐱. How can I help you?"

    elif "your name" in user_input:
        return "My name is Kitty, your AI assistant."

    elif "how are you" in user_input:
        return "I'm doing great! Thanks for asking 😊"

    elif re.search(r"\bbye\b", user_input):
        return "Goodbye! Talk to you soon 👋"

    else:
        return "I'm still learning. Can you ask something else?"
```

## Try these messages

The whole-word checks prevent `hi` from matching inside `this` or `history`. The goodbye rule currently matches standalone `bye`, not `goodbye`.

```python
test_messages = [
    "hi",
    "HELLO",
    "this is interesting",
    "history is important",
    "what is your name?",
    "how are you?",
    "bye",
]

for message in test_messages:
    print(f"You: {message}")
    print(f"Kitty: {kitty_chat(message)}")
    print("-" * 40)
```

## Interactive chat (optional)

Run this cell in a terminal or a notebook environment that supports `input()`. Type `exit` to stop.

```python
while True:
    user = input("You: ")
    if user.lower().strip() == "exit":
        print("Kitty: Bye! 👋")
        break
    print("Kitty:", kitty_chat(user))
```
